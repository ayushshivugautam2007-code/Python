```python
#Exercise-1
def safe_divide(numerator, denominator):
    try:
        return numerator / denominator
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None

# User input and function call
num = float(input("Enter numerator: "))
den = float(input("Enter denominator: "))

result = safe_divide(num, den)
print("Result:", result)
```

```python
#Exercise-2
import math

def calculate_square_root(number1):
    try:
        if number1 < 0:
            raise ValueError("Negative number")
        return math.sqrt(number1)
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None

# User input and function call
val = float(input("Enter a number: "))
sqrt_result = calculate_square_root(val)
if sqrt_result is not None:
    print("Square Root:", sqrt_result)
```

```python
#Exercise-3
def complex_calculation(num):
    try:
        result = num / (num - 5)
        return result
    except Exception as e:
        print("An error occurred during calculation.")
        return None

# User input and function call
num_input = float(input("Enter a number: "))
calc_result = complex_calculation(num_input)
if calc_result is not None:
    print("Result:", calc_result)
```
