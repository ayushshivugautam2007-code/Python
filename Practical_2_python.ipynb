```python
shopping_list = []
print("Initiaally, the shopping list is: ")
print(shopping_list)
```

```python
shopping_list.append("Watch")
shopping_list.append("Laptop")
shopping_list.append("Shoes")
shopping_list.append("Pen")
shopping_list.append("Clothes")
print("\n Shooping List after adding  items: ")
print(shopping_list)
```

```python
shopping_list.append("Football")
print("\n Shopping list after adding Football: ")
print(shopping_list)
```

```python
last_item = shopping_list[-1]
print("\n Last item from the hopping list: ")
print(last_item)
```

```python
print("\n Entire shopping list: ")
for item in shopping_list:

  print(item)
```

```python
print("\n Important items: ")
print("1.", shopping_list[1])
print("2.", shopping_list[2])
```

```python
print("\n Before changing  Pen: ")
print(shopping_list)
shopping_list[3] = "Notebook"
print("\n After changing  pen to notebook: ")
print(shopping_list)
```

```python
if "Watch" in shopping_list: shopping_list.remove("Shoes")
else: print("'Watch' is not in the shopping list.")
print("\n Final shopping list after deleting the least important items: ")
for item in shopping_list: print(item)
print("\n Total number of items in the shopping list: ")
print(len(shopping_list))
```
