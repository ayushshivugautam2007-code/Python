```python
# 1. Create an empty dictionary
inventory = {}

# 2. Store the first product details in variables
ProductNo1_name = "Mobile Phone"
ProductNo1_quantity = 5
ProductNo1_price = 20000
ProductNo1_releaseYear = 2020

# 3. Add the first product details to inventory
inventory["ProductNo1_name"] = ProductNo1_name
inventory["ProductNo1_quantity"] = ProductNo1_quantity
inventory["ProductNo1_price"] = ProductNo1_price
inventory["ProductNo1_releaseYear"] = ProductNo1_releaseYear

# 4. Store the second product details in variables
ProductNo2_name = "Laptop"
ProductNo2_quantity = 3
ProductNo2_price = 50000
ProductNo2_releaseYear = 2022

# 5. Add the second product details to inventory
inventory["ProductNo2_name"] = ProductNo2_name
inventory["ProductNo2_quantity"] = ProductNo2_quantity
inventory["ProductNo2_price"] = ProductNo2_price
inventory["ProductNo2_releaseYear"] = ProductNo2_releaseYear

# 6. Display the products present in inventory
print("Products present in inventory:")
print(inventory)

# 7. Check if release years are present in inventory
print("\nChecking release years:")

if "ProductNo1_releaseYear" in inventory:
    print("ProductNo1_releaseYear is present in inventory.")
else:
    print("ProductNo1_releaseYear is not present in inventory.")

if "ProductNo2_releaseYear" in inventory:
    print("ProductNo2_releaseYear is present in inventory.")
else:
    print("ProductNo2_releaseYear is not present in inventory.")

# 8. Delete release year of both products
del inventory["ProductNo1_releaseYear"]
del inventory["ProductNo2_releaseYear"]

print("\nInventory after deleting release years:")
print(inventory)
```

```python
# Create a tuple of product prices
prices = (250, 300, 150, 400, 100, 350, 200)

# Find and print the highest price
print("Highest price:", max(prices))

# Find and print the lowest price
print("Lowest price:", min(prices))

# Calculate and print the total sum
print("Total sum:", sum(prices))

# Convert tuple into a sorted list
sorted_prices = sorted(prices)
print("Sorted prices:", sorted_prices)

# Try to modify an element in the tuple
try:
    prices[0] = 500
except TypeError as e:
    print("Error:", e)
```
