```python
def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            s1 = strings[i]
            s2 = strings[i + 1]

            result = ""
            for a, b in zip(s1, s2):
                result += a + b

            file.write(result + "\n")


strings = ["abc", "123", "xyz", "456"]

zipper_file("output.txt", strings)

print("File created successfully!")
```

```python
def reformat_date(dates):

    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in dates:
        month, day, year = date.split("/")

        sentence = "Office Hours on " + months[month] + " " + day + ", " + year

        result.append(sentence)

    return result


dates = ["01/15/2026", "05/09/2026", "12/25/2026"]

print(reformat_date(dates))
```
