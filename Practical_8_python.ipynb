```python
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan',
             'Neha', 'Arjun', 'Simran', 'Vikas', 'Pooja'],
    'Age': [25, 32, 28, 40, 35, 22, 45, 31, 27, 38],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2000, 3500, 1500, 4000, 3000,
                        2500, 5000, 1800, 2200, 3200]
}

df2 = pd.DataFrame(data)

print(df2.head())

print(df2.describe())

print(df2.isnull().sum())

df2['Age'] = df2['Age'].fillna(df2['Age'].median())
df2['Purchase_Amount'] = df2['Purchase_Amount'].fillna(
    df2['Purchase_Amount'].median()
)

print(df2)
```

```python
result = df2[df2['Age'] > 30]
print(result)
```

```python
total = df2.groupby('Gender')['Purchase_Amount'].sum()
print(total)
```

```python
selected = df2[['Customer_ID', 'Name', 'Purchase_Amount']]
print(selected)
```

```python
df2['Discounted_Price'] = df2['Purchase_Amount'] * 0.90

print(df2)
```

```python
average = df2.groupby('Gender')['Purchase_Amount'].mean()
print(average)
```

```python
sorted_df = df2.sort_values(
    by='Purchase_Amount', ascending=False
)

print(sorted_df)
```
