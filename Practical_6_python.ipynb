```python
#Problem-1
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print(self.name, self.id)

class Emp(Person):
    def __init__(self, name, id):
        # Call parent class constructor using super()
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")

# Object instantiation and method calling
Emp_details = Emp("Mayank", 103)
Emp_details.Display()
Emp_details.Print()
```

```python
#Problem-2
def is_anagram(str1, str2):
    # Normalize by lowercasing and removing spaces
    s1 = str1.replace(" ", "").lower()
    s2 = str2.replace(" ", "").lower()
    return sorted(s1) == sorted(s2)

# Driver test
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

if is_anagram(s1, s2):
    print(f"'{s1}' and '{s2}' are anagrams.")
else:
    print(f"'{s1}' and '{s2}' are not anagrams.")
```

```python
#Problem-3
def is_armstrong(number):
    num_str = str(number)
    power = len(num_str)
    total = sum(int(digit) ** power for digit in num_str)
    return total == number

# Driver test
num = int(input("Enter a number: "))

if is_armstrong(num):
    print(f"{num} is an Armstrong number.")
else:
    print(f"{num} is not an Armstrong number.")
```

```python
#Problem-4
class Vehicle:
    # Task-2: Default color for all vehicles
    color = "white"

    # Task-1: Attributes for max speed and mileage
    def __init__(self, max_speed, mileage, seating_capacity=None):
        self.max_speed = max_speed
        self.mileage = mileage
        self.seating_capacity = seating_capacity

    # Task-3: Method to assign seating capacity
    def set_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    # Task-4: Method to display all properties of the vehicle
    def display_properties(self):
        print(f"Color: {self.color}")
        print(f"Max Speed: {self.max_speed} kmph")
        print(f"Mileage: {self.mileage} kmpl")
        print(f"Seating Capacity: {self.seating_capacity}")
        print("-" * 30)

# Task-5: Create the two vehicle objects as specified
car1 = Vehicle(max_speed=200, mileage=20)
car1.set_seating_capacity(5)

car2 = Vehicle(max_speed=180, mileage=25)
car2.set_seating_capacity(4)

# Displaying details of both objects
print("Vehicle 1 Details:")
car1.display_properties()

print("Vehicle 2 Details:")
car2.display_properties()
```
